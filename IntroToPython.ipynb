```python
from operator import itemgetter

# Example array
my_array = [[1, 5, 2], [3, 1, 8], [7, 9, 0]]

print("Original array:", my_array)

# Sort the array by the last element of each sub-array
sorted_array = sorted(my_array, key=itemgetter(-1))

print("Sorted array by last element:", sorted_array)
```
